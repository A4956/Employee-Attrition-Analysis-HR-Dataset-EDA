# **Which department has the highest number of employees**

```python
df['Department'].value_counts()
```

```python
count= df['Department'].value_counts()
ax= count.plot(kind='bar', color='lightgreen')
ax.bar_label(ax.containers[0],padding=3)



plt.title("Department with highest number of employees")
plt.xlabel("Department")
plt.ylabel("Number of employess")
plt.show()
```

Production has the highest employee count

## What is the average salary distribution across employees?

```python
sns.histplot(df['Salary'], color="lightgreen")
plt.title("Salary Distribution")
plt.xlabel("Salary")
plt.ylabel("Frequency")
plt.show()
```

```python
df['Salary'].plot(kind='hist')
```


Inight:
